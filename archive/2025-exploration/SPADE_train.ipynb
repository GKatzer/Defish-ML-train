```python
import os
import numpy as np
from PIL import Image

test_img_path = r"F:\Datasets\Reddit-fishes\aquarium_new\mvtec\fish\test\not-a-fish"
mask_dir = r"F:\Datasets\Reddit-fishes\aquarium_new\mvtec\fish\ground_truth\not-a-fish"

os.makedirs(mask_dir, exist_ok=True)

for img_name in os.listdir(test_img_path):
    if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
        # Создаём пустую чёрную маску (256x256 пикселей)
        mask = np.zeros((256, 256), dtype=np.uint8)
        # Сохраняем с тем же именем, но с суффиксом _mask
        mask_name = os.path.splitext(img_name)[0] + "_mask.png"
        Image.fromarray(mask).save(os.path.join(mask_dir, mask_name))
```
