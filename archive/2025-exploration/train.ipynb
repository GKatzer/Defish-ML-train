```python
from pathlib import Path
from collections import Counter

label_dir = Path("Dataset_hand_label/valid/labels")  # или train
class_counter = Counter()
empty_count = 0

for txt_path in label_dir.glob("*.txt"):
    with open(txt_path, "r") as f:
        lines = [line.strip() for line in f if line.strip()]
    
    if not lines:
        empty_count += 1
        print(f"[EMPTY] {txt_path.resolve()}")
    else:
        for line in lines:
            class_id = int(line.split()[0])
            class_counter[class_id] += 1

print("\n📊 Объекты по классам:")
for cls, count in class_counter.items():
    print(f"Class {cls}: {count} объектов")

print(f"\n🧼 Пустых файлов: {empty_count}")
```
