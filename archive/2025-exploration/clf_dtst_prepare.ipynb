```python
import os
from PIL import Image
import cv2
from ultralytics import YOLO

# Paths
model_path = r".\runs\detect\fish_easy_aug_n5\weights\best.pt"
input_dir = "./All_imgs"
output_dir = "./cropped_fish"
os.makedirs(output_dir, exist_ok=True)

# Load YOLO model
model = YOLO(model_path)

# Loop over all files
for img_name in os.listdir(input_dir):
    input_path = os.path.join(input_dir, img_name)

    # Convert to .jpg if format is not supported by OpenCV
    if not img_name.lower().endswith((".jpg", ".jpeg", ".png")):
        try:
            img = Image.open(input_path).convert("RGB")
            converted_name = os.path.splitext(img_name)[0] + ".jpg"
            input_path = os.path.join(input_dir, converted_name)
            img.save(input_path)
        except Exception as e:
            print(f"[SKIP] Can't convert {img_name}: {e}")
            continue

    # Load image with OpenCV
    img = cv2.imread(input_path)
    if img is None:
        print(f"[SKIP] Failed to read: {input_path}")
        continue

    # Predict
    results = model.predict(source=img, save=False, conf=0.2)[0]

    # Crop each detection
    for i, box in enumerate(results.boxes.xyxy.cpu().numpy()):
        x1, y1, x2, y2 = map(int, box)
        crop = img[y1:y2, x1:x2]

        # Skip if the crop is empty
        if crop.size == 0:
            continue

        out_path = os.path.join(output_dir, f"{os.path.splitext(img_name)[0]}_crop{i}.jpg")
        cv2.imwrite(out_path, crop)

print("✅ Done cropping.")
```
